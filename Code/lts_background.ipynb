## The Problem: Why Most People Don't Cycle

In cities across the UK and Ireland, cycling accounts for a tiny fraction of
commuting trips — despite being faster than driving for many short journeys,
cheaper than public transport, and better for health. The most commonly cited
barrier is not distance, weather, or fitness. It is **fear of traffic**.

Survey after survey finds the same result: the majority of the population would
consider cycling *if* they felt safe on the roads. But "feeling safe" is
subjective, and traditional transport planning metrics — vehicle throughput,
average speed, junction capacity — tell us nothing about it.

This is the gap that **Level of Traffic Stress (LTS)** fills.

## What is Level of Traffic Stress?

Level of Traffic Stress (LTS) is a classification framework introduced by
Mekuria, Furth & Nixon (2012) that rates every road segment on a scale of
**1 to 4** based on how stressful it would feel to cycle on. The framework
was originally developed at the Mineta Transportation Institute in San José,
California, and has since been applied in cities across North America, Europe,
and Australasia.

The key insight is that LTS does not measure *danger* (crash risk) or
*difficulty* (gradient, surface quality). It measures **perceived comfort** —
the level of traffic interaction that a cyclist must tolerate. A quiet
residential street with no cycle lane might be LTS 1 (very comfortable),
while a wide road with a painted cycle lane alongside fast traffic might
be LTS 3 (stressful for most people).

## The Four Levels

Each LTS level corresponds to a different **population tolerance group** —
the segment of the population willing to cycle under those conditions.

| LTS | Label | Who Would Ride Here? | Road Characteristics | Example |
|:---:|:------|:---------------------|:---------------------|:--------|
| **1** | **Low Stress** | Children, elderly, and the least confident cyclists. Suitable for all ages and abilities. | Quiet residential streets, off-road paths, physically separated cycle tracks. Speeds ≤ 20 mph, minimal motor traffic. | A tree-lined residential cul-de-sac; a canal towpath; a fully kerb-separated cycle lane. |
| **2** | **Moderate Stress** | The "interested but concerned" majority — the mainstream adult population who would cycle if conditions felt safe. | Roads with moderate traffic, speeds ≤ 25 mph, ≤ 3 lanes. May have painted cycle lanes. Limited mixing with motor vehicles. | A neighbourhood collector road with a marked cycle lane; a 20 mph high street with low traffic. |
| **3** | **High Stress** | "Enthused and confident" cyclists — people who cycle regularly and are comfortable mixing with traffic in most conditions. | Multi-lane roads, speeds 25–40 mph, or roads without cycling infrastructure on busy corridors. Frequent interaction with turning vehicles. | A dual carriageway with an advisory cycle lane; a busy suburban arterial with on-street parking. |
| **4** | **Very High Stress** | Only the "strong and fearless" — a tiny minority comfortable on any road. | High-speed roads (> 40 mph), 4+ lanes, no cycling provision. Heavy goods vehicle traffic. Hostile junctions. | An urban motorway slip road; a national speed limit dual carriageway; a multi-lane roundabout. |

The population segments are based on Roger Geller's widely cited "Four Types
of Cyclists" typology (Portland Bureau of Transportation, 2006), which estimates
that roughly **60% of the population** falls into the "interested but concerned"
group — people who *want* to cycle but are deterred by traffic stress. LTS 2 is
the threshold that captures this majority.

## How is LTS Determined?

The original Mekuria et al. (2012) framework uses a decision matrix based on
three primary road characteristics:

1. **Speed limit** — the posted or implied maximum speed of motor traffic
2. **Number of lanes** — total through-lanes in the direction of travel
3. **Cycling infrastructure** — the presence and type of cycle facility

These three variables interact to produce the LTS assignment:

### Roads Without Cycling Infrastructure

| | ≤ 25 mph | 26–30 mph | > 30 mph |
|:---|:---:|:---:|:---:|
| **≤ 2 lanes** | LTS 2 | LTS 3 | LTS 4 |
| **3 lanes** | LTS 2 | LTS 3 | LTS 4 |
| **≥ 4 lanes** | LTS 3 | LTS 4 | LTS 4 |

### Roads With a Painted (Unprotected) Cycle Lane

| | ≤ 25 mph | 26–30 mph | > 30 mph |
|:---|:---:|:---:|:---:|
| **≤ 2 lanes** | LTS 2 | LTS 2 | LTS 3 |
| **3 lanes** | LTS 2 | LTS 3 | LTS 3 |
| **≥ 4 lanes** | LTS 3 | LTS 3 | LTS 4 |

### Roads With a Protected (Physically Separated) Cycle Track

| | Any speed | Any lanes |
|:---|:---:|:---:|
| **Kerb-separated or off-road** | LTS 1 | LTS 1 |
| **Adjacent track (e.g. stepped)** | LTS 2 | LTS 2 |

Several additional rules handle special cases:

- **Residential streets** (highway = residential, living_street) → LTS 1
- **Off-road paths** (cycleway, path with bicycle access) → LTS 1
- **Motor-vehicle-prohibited roads** → LTS 1
- **Unsignalised intersections** propagate the highest LTS of any adjoining
  edge to all edges at that vertex, reflecting the stress of crossing
  without signal protection

## Why LTS Matters: The Cyclist's Perspective

### 1. It Measures What Cyclists Actually Care About

Traditional road safety metrics count crashes and casualties. These are
important, but they suffer from a fundamental bias: **if a road is so
intimidating that nobody cycles on it, it will have zero cycling casualties
and appear "safe"** in the statistics. LTS captures the *deterrent effect*
of hostile infrastructure — the trips that never happen because people
choose to drive instead.

### 2. It Reveals Network Gaps, Not Just Segment Quality

A city might have hundreds of kilometres of quiet residential streets (LTS 1)
and even some dedicated cycle tracks. But if these low-stress segments are
disconnected — separated by even a single high-stress junction or arterial
crossing — the network is useless for most people. LTS combined with
connectivity analysis (Mekuria's *Percent Trips Connected*) reveals these
critical gaps.

Consider this analogy: a pedestrian network with excellent pavements on
every residential street, but no safe crossing at any main road. Each
individual pavement is fine; the network is broken.

### 3. It Directly Maps to Policy Interventions

Because LTS is determined by speed, lanes, and infrastructure, each
classification implies a specific set of interventions to downgrade it:

| Current LTS | Target LTS | Possible Interventions |
|:---:|:---:|:---|
| 4 → | 3 | Reduce speed limit; add advisory cycle lane |
| 4 → | 2 | Reduce speed limit *and* add mandatory cycle lane |
| 3 → | 2 | Add protected cycle lane or reduce to 20 mph |
| 3 → | 1 | Add fully separated cycle track |
| 2 → | 1 | Implement modal filter or reduce to living street |

This makes LTS an actionable planning tool: planners can identify the
specific bottleneck links that, if downgraded, would connect the largest
number of origin-destination pairs at a lower stress level.


## From Segments to Networks: Percent Trips Connected

LTS classification alone tells us the stress level of individual road
segments. But the real question is: **can you get from A to B using only
low-stress roads?**

Mekuria et al. (2012) proposed two network-level connectivity metrics:

| Metric | Definition |
|:---|:---|
| **Percent Trips Connected** | The share of commuting demand (weighted by trip count) that can be served by a path using only edges ≤ a given LTS, without exceeding a 25% detour compared to the shortest route. |
| **Percent Nodes Connected** | The share of origin-destination pairs (unweighted) satisfying the same criterion. |

The 25% detour threshold (or 0.33 miles absolute for short trips) ensures
that the low-stress route is a *practical* alternative, not an absurd
circumnavigation.

These metrics transform LTS from a per-segment observation into a
**network-level diagnostic**. A city might have 60% of its road segments
at LTS 1 — but if only 2% of trips can be completed at LTS ≤ 2 without
undue detour, the low-stress network is functionally disconnected.

## Implementation in This Project

In this project, I apply the LTS framework to Belfast using the following
pipeline:

| Step | Method | Data Source |
|:---|:---|:---|
| Road network extraction | pyrosm (PBF, full tag preservation) + OSMnx (graph topology) | OpenStreetMap |
| Cycle infrastructure classification | 6-category scheme following Jeong & Smith (2025) Table 1 | OSM `cycleway`, `bicycle`, `highway` tags |
| Per-edge LTS classification | Augmented Conveyal scheme with speed × lane matrix | OSM `maxspeed`, `lanes` tags |
| Infrastructure override | Protected facilities → LTS 1/2 regardless of road class | Cycle infrastructure classification |
| Fallback for missing data | Routino highway-type lookup for unclassified edges | OSM `highway` tag |
| Intersection adjustment | Jeong & Smith (2025) signalised junction modification | OSM traffic signal nodes |
| Connectivity analysis | Mekuria et al. (2012) Percent Trips/Nodes Connected | Census 2021 OD flows (SDZ level) |
